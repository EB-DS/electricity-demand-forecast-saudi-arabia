## 3. Data Preparation and Feature Engineering

### Data Loading and Initial Inspection

The analysis begins by loading a static snapshot of the dataset *Saudi Arabia Electricity Load Monthly by Region*, downloaded directly from the KAPSARC Energy Data Portal and stored locally to ensure reproducibility. Using a locally stored file avoids potential inconsistencies arising from API updates or metadata revisions and is particularly appropriate given the discontinued status of the dataset.

Upon ingestion, an initial inspection is conducted to verify the dataset’s structure, variable types, temporal coverage, and regional identifiers.
```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("data/saudi-arabia-electricity-load-monthly-by-region.xlsx")

df = pd.read_excel(DATA_PATH)
df.head()
```

### Data Cleaning and Reshaping

The dataset is reshaped into a panel time series format with explicit datetime indexing to support temporal analysis and machine learning workflows.


```python
import pandas as pd

# Standardize column names for clarity and consistency
df = df.rename(columns={
    "Date": "date_raw",
    "Indicator": "indicator",
    "Value": "value"
}).copy()

# Convert monthly period strings (YYYY-MM) to datetime format
df["date"] = pd.to_datetime(
    df["date_raw"].astype(str),
    format="%Y-%m",
    errors="coerce"
)

# Extract region and load type from the indicator field
df["region"] = (
    df["indicator"]
    .str.extract(r"^(.+?)\s+region", expand=False)
    .str.strip()
)

df["load_type"] = df["indicator"].str.extract(
    r"(Maximum|Average|Minimum)",
    expand=False
)

# Remove records with incomplete or invalid fields
df = df.dropna(subset=["date", "region", "load_type", "value"]).copy()

# Ensure load values are numeric
df["value"] = pd.to_numeric(df["value"], errors="coerce")
df = df.dropna(subset=["value"]).copy()

# Reshape data into a panel structure with separate load measures
df_wide = (
    df.pivot_table(
        index=["region", "date"],
        columns="load_type",
        values="value",
        aggfunc="mean"
    )
    .reset_index()
)

# Apply consistent naming conventions to load variables
df_wide = df_wide.rename(columns={
    "Maximum": "load_max_gw",
    "Average": "load_avg_gw",
    "Minimum": "load_min_gw"
})

# Sort observations and set a multi-index for time series analysis
df_wide = df_wide.sort_values(["region", "date"])
df_wide = df_wide.set_index(["region", "date"])

# Preview the cleaned and reshaped dataset
df_wide.head()
```

### Handling Missing Values

The dataset was evaluated for missing observations at both the variable and regional levels following reshaping into a panel time series structure. No missing values were detected across any electricity load measures or regions over the study period.

Given the completeness of the data, no imputation or data exclusion procedures were required. All observations were retained for subsequent analysis, ensuring consistent temporal coverage and preserving the integrity of the original measurements.

```python
# Construct a complete (region × month) index to verify consistent monthly coverage
regions = df_wide.index.get_level_values("region").unique()
start_month = df_wide.index.get_level_values("date").min()
end_month = df_wide.index.get_level_values("date").max()
monthly_index = pd.date_range(start_month, end_month, freq="MS")

complete_index = pd.MultiIndex.from_product(
    [regions, monthly_index],
    names=["region", "date"]
)

df_complete = (
    df_wide
    .reindex(complete_index)
    .sort_index()
)

# Compare the observed and expected number of observations
rows_observed = df_wide.shape[0]
rows_expected = df_complete.shape[0]
rows_observed, rows_expected
```

### Feature Engineering: Seasonality and Lag Features

Electricity demand exhibits strong temporal structure driven by seasonal weather patterns, economic activity, and recurrent institutional behaviors. To capture these dynamics, additional time-based and autoregressive features are constructed to support both classical time series analysis and machine learning models. Seasonal effects are represented using calendar-based variables derived from the monthly timestamp. These include the month of the year to account for intra-annual variation in electricity consumption, as well as a binary indicator distinguishing summer months, which are associated with elevated cooling demand in Saudi Arabia.

In addition, lagged demand features are introduced to capture short- and medium-term temporal dependence. These lag variables allow models to learn persistence, momentum, and delayed adjustment effects that are characteristic of electricity load behavior across regions. All features are engineered at the regional level to preserve spatial heterogeneity and prevent information leakage across regions.

```python
# Create calendar-based seasonal features
df_features = df_complete.copy()

df_features["month"] = (
    df_features
    .index
    .get_level_values("date")
    .month
)

# Indicator for summer months (May–September)
df_features["is_summer"] = df_features["month"].isin([5, 6, 7, 8, 9]).astype(int)

# Generate lagged load features by region
LAG_PERIODS = [1, 3, 6]

for lag in LAG_PERIODS:
    df_features[f"load_avg_lag_{lag}"] = (
        df_features
        .groupby(level="region")["load_avg_gw"]
        .shift(lag)
    )
    df_features[f"load_max_lag_{lag}"] = (
        df_features
        .groupby(level="region")["load_max_gw"]
        .shift(lag)
    )
    df_features[f"load_min_lag_{lag}"] = (
        df_features
        .groupby(level="region")["load_min_gw"]
        .shift(lag)
    )

# Display the feature-enhanced dataset
df_features.head()
```

**Note on lagged features**

Lagged features introduce structurally missing values in the initial months of each regional time series. These observations are retained during feature construction and will be excluded only at the modeling stage to preserve temporal integrity and prevent information leakage.

```python
from __future__ import annotations

from pathlib import Path
import re

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
import requests
from shapely.geometry import Point


DATA_PATH = Path(
    r"C:\Users"
    r"\saudi-arabia-electricity-load-monthly-by-region.xlsx"
)

NE_URL = (
    "https://naturalearth.s3.amazonaws.com/110m_cultural/"
    "ne_110m_admin_0_countries.zip"
)
NE_ZIP_PATH = Path("data/ne_110m_admin_0_countries.zip")


def download_file(url: str, dest: Path, chunk_size: int = 1024 * 1024) -> Path:
    dest.parent.mkdir(parents=True, exist_ok=True)
    if dest.exists():
        return dest

    resp = requests.get(url, stream=True, timeout=60)
    resp.raise_for_status()

    with dest.open("wb") as f:
        for chunk in resp.iter_content(chunk_size=chunk_size):
            if chunk:
                f.write(chunk)

    return dest


def first_existing_column(df: pd.DataFrame, candidates: list[str]) -> str:
    for col in candidates:
        if col in df.columns:
            return col
    raise ValueError(f"None of these columns exist: {candidates}. Found: {list(df.columns)}")


def load_saudi_boundary() -> gpd.GeoDataFrame:
    zip_path = download_file(NE_URL, NE_ZIP_PATH)
    world = gpd.read_file(zip_path)
    iso_col = first_existing_column(world, ["ISO_A3", "ADM0_A3", "SOV_A3", "GU_A3"])
    saudi = world.loc[world[iso_col].astype(str).str.upper().eq("SAU")].copy()
    if saudi.empty:
        raise ValueError("Saudi Arabia polygon not found using ISO code 'SAU'.")
    return saudi.to_crs(epsg=3857)


def build_region_points() -> gpd.GeoDataFrame:
    points = gpd.GeoDataFrame(
        {
            "Region": ["Central", "Eastern", "Western", "Southern"],
            "City": ["Riyadh", "Dammam", "Jeddah", "Abha"],
        },
        geometry=[
            Point(46.6753, 24.7136),
            Point(50.1033, 26.4207),
            Point(39.1925, 21.4858),
            Point(42.5117, 18.2164),
        ],
        crs="EPSG:4326",
    )
    return points.to_crs(epsg=3857)


df = pd.read_excel(DATA_PATH)

required_cols = {"Indicator", "Value"}
missing = required_cols - set(df.columns)
if missing:
    raise ValueError(f"Dataset is missing required columns: {sorted(missing)}")

indicator = df["Indicator"].astype(str)

m_region = indicator.str.extract(r"^(?P<Region>.*?)\s+region\s+-", flags=re.IGNORECASE)
m_stat = indicator.str.extract(r"region\s+-\s*(?P<Stat>Average|Maximum|Minimum)", flags=re.IGNORECASE)

parsed = df.copy()
parsed["Region"] = m_region["Region"].str.strip().str.title()
parsed["Stat"] = m_stat["Stat"].str.strip().str.title()
parsed["Value"] = pd.to_numeric(parsed["Value"], errors="coerce")

region_load = (
    parsed.loc[parsed["Stat"].eq("Average") & parsed["Region"].notna(), ["Region", "Value"]]
    .groupby("Region", as_index=False)["Value"]
    .mean()
    .rename(columns={"Value": "Avg Load (GW)"})
)

expected_regions = {"Central", "Eastern", "Western", "Southern"}
found_regions = set(region_load["Region"].unique())
missing_regions = sorted(expected_regions - found_regions)
if missing_regions:
    raise ValueError(f"Could not compute average load for region(s): {missing_regions}. Check Indicator parsing.")

saudi = load_saudi_boundary()
region_points = build_region_points()

region_points = region_points.merge(region_load, on="Region", how="left")
unmatched = region_points.loc[region_points["Avg Load (GW)"].isna(), "Region"].tolist()
if unmatched:
    raise ValueError(f"Missing load values for region(s): {unmatched}.")

region_colors = {
    "Central": "#1f77b4",
    "Eastern": "#ff7f0e",
    "Western": "#2ca02c",
    "Southern": "#d62728",
}

fig, ax = plt.subplots(figsize=(9, 7.5))
ax.set_facecolor("#f5f7fb")

saudi.plot(
    ax=ax,
    edgecolor="#1f2937",
    facecolor="#dbeafe",
    linewidth=1.2,
    zorder=1,
)

for region, color in region_colors.items():
    subset = region_points[region_points["Region"] == region]
    subset.plot(
        ax=ax,
        color=color,
        markersize=95,
        zorder=3,
        label=region,
        edgecolor="white",
        linewidth=1.2,
    )

minx, miny, maxx, maxy = saudi.total_bounds
dx = (maxx - minx) * 0.02
dy = (maxy - miny) * 0.02

for geom, region, city, avg_load in zip(
    region_points.geometry,
    region_points["Region"],
    region_points["City"],
    region_points["Avg Load (GW)"],
):
    ax.text(
        geom.x + dx,
        geom.y + dy,
        f"{region} ({city})\nAvg Load: {avg_load:.1f} GW",
        fontsize=9,
        bbox=dict(
            boxstyle="round,pad=0.25",
            facecolor="white",
            edgecolor="none",
            alpha=0.9,
        ),
        zorder=4,
    )

ax.set_title(
    "Saudi Arabia Regional Reference Map - Average Monthly Electricity Load (2009–2019)",
    fontsize=13,
    pad=18,
)
ax.legend(title="Region", loc="upper right", frameon=True, borderpad=0.8, labelspacing=0.6)

ax.set_xlim(minx - (maxx - minx) * 0.06, maxx + (maxx - minx) * 0.06)
ax.set_ylim(miny - (maxy - miny) * 0.06, maxy + (maxy - miny) * 0.06)

ax.set_axis_off()

ax.text(
    0.5,
    0.02,
    "Boundary source: Natural Earth (Admin 0 – Countries, 110m).",
    transform=ax.transAxes,
    ha="center",
    va="bottom",
    fontsize=8,
    color="#374151",
)

plt.tight_layout()
plt.show()
```

###  Statistical Baseline: Seasonal ARIMA (SARIMA)

Seasonal Autoregressive Integrated Moving Average (SARIMA) models serve as a widely used benchmark in electricity demand forecasting. These models represent the current value of a time series as a function of its own past values, past forecast errors, and recurring seasonal patterns.

In intuitive terms:
- **Autoregressive components** capture persistence in electricity demand across consecutive months.
- **Moving average components** account for short-term shocks and unexplained fluctuations.
- **Seasonal components** explicitly model recurring annual demand cycles, such as elevated summer consumption driven by cooling needs.

SARIMA models are particularly valuable due to their transparency and interpretability. Model parameters can be directly linked to underlying temporal dynamics, making SARIMA a useful reference point for understanding baseline demand behavior. In this study, SARIMA models are estimated separately for each region to capture region-specific demand persistence and seasonal intensity.


```python
import warnings
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Suppress non-critical statsmodels warnings for a clean public notebook
warnings.filterwarnings("ignore", category=UserWarning)

# Select regional training series and assign explicit monthly frequency
central_train = (
    train_data
    .xs("Central", level="region")["load_avg_gw"]
    .asfreq("MS")
)

# Estimate SARIMA model with annual seasonality (monthly data)
sarima_model = SARIMAX(
    central_train,
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

sarima_results = sarima_model.fit(disp=False)
sarima_results.summary()
```

**Interpretation**

The SARIMA model captures key temporal dynamics in regional electricity demand, including short-term persistence and annual seasonality. The autoregressive term is statistically significant, indicating meaningful month-to-month dependence in electricity load. The seasonal moving average component is also significant, confirming the presence of recurring annual patterns consistent with climate-driven demand cycles. Several parameters exhibit large standard errors, reflecting the relatively limited sample size and the inherent volatility of electricity demand at the regional level. Diagnostic statistics, including the Ljung–Box test and residual normality measures, do not indicate strong autocorrelation or distributional violations, suggesting an adequate baseline fit.

This model is treated as a statistical benchmark rather than a final forecasting solution. Its primary role is to provide an interpretable reference point against which more flexible machine learning models can be evaluated.


### Machine Learning Approach: XGBoost

While SARIMA models effectively capture linear temporal dependence and fixed seasonal patterns, electricity demand may also be shaped by nonlinear relationships and interactions that are difficult to model parametrically. To address this limitation, a machine learning approach based on Extreme Gradient Boosting (XGBoost) is employed.

XGBoost is an ensemble learning algorithm that constructs a sequence of decision trees, where each successive tree focuses on correcting the errors of previous ones. In a time-series forecasting context, temporal structure is incorporated through engineered features rather than explicit time indices. These features include:
- Lagged values of electricity load.
- Calendar-based indicators capturing seasonal effects.
- Region-specific patterns modeled through separate regional estimations.

The primary strength of XGBoost lies in its flexibility. The model can learn nonlinear relationships, asymmetric seasonal responses, and complex interactions among predictors without imposing strong assumptions about the data-generating process. This makes it well suited for modeling evolving electricity demand behavior, particularly in the presence of structural change or heterogeneous regional dynamics.


```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import matplotlib.pyplot as plt

# Select feature set
feature_cols = [
    "month",
    "is_summer",
    "load_avg_lag_1",
    "load_avg_lag_3",
    "load_avg_lag_6"
]

# Drop initial rows with NaNs from lag construction
train_ml = train_data.dropna(subset=feature_cols)
val_ml = val_data.dropna(subset=feature_cols)

X_train = train_ml[feature_cols]
y_train = train_ml["load_avg_gw"]

X_val = val_ml[feature_cols]
y_val = val_ml["load_avg_gw"]

# Initialize XGBoost model
xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

# Train model
xgb_model.fit(X_train, y_train)

# Validation prediction
val_preds = xgb_model.predict(X_val)

# --- Display validation performance (output) ---
mae = mean_absolute_error(y_val, val_preds)
rmse = np.sqrt(mean_squared_error(y_val, val_preds))

print(f"XGBoost Validation MAE: {mae:.3f} GW")
print(f"XGBoost Validation RMSE: {rmse:.3f} GW")
```

```python
region_name = "Central"

# Subset validation data for one region
val_r = val_ml.xs(region_name, level="region")

# Actual and predicted values
y_r = val_r["load_avg_gw"]
pred_r = xgb_model.predict(val_r[feature_cols])

# Date axis
x_axis = pd.to_datetime(val_r.index)

# Plot
plt.figure(figsize=(10, 4))

plt.plot(
    x_axis,
    y_r.values,
    label="Actual",
    color="blue",
    linewidth=2
)

plt.plot(
    x_axis,
    pred_r,
    label="Predicted",
    color="red",
    linestyle="--",
    linewidth=2
)

plt.title(f"XGBoost Validation: Actual vs Predicted ({region_name})")
plt.xlabel("Date")
plt.ylabel("Average Load (GW)")
plt.legend()
plt.tight_layout()
plt.show()
```

**Note:** This validation plot focuses on the Central region to illustrate the XGBoost model’s out-of-sample performance at the regional level. Other regions are evaluated separately to account for differences in demand scale and seasonal intensity.

The quantitative metrics used to evaluate forecasting performance. Forecast accuracy metrics provide an objective measure of how closely model predictions align with observed electricity demand. Two commonly used metrics are employed: Root Mean Squared Error (RMSE) and Mean Absolute Percentage Error (MAPE).

RMSE measures the average magnitude of forecast errors in the same units as electricity load, placing greater emphasis on large deviations. MAPE expresses forecast errors as a percentage of observed values, allowing performance comparison across regions with different demand scales.


```python
# Reconstruct test series for Central region
central_test = (
    test_data
    .xs("Central", level="region")["load_avg_gw"]
    .asfreq("MS")
)

# Generate SARIMA forecasts for the test period
sarima_forecast = sarima_results.get_forecast(steps=len(central_test))
sarima_pred = sarima_forecast.predicted_mean

# Define accuracy metrics
def rmse(y_true, y_pred):
    return np.sqrt(np.mean((y_true - y_pred) ** 2))

def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

# Create comparison DataFrame
comparison_df = pd.DataFrame({
    "Actual": central_test.values,
    "SARIMA_Forecast": sarima_pred.values
}, index=central_test.index)

# Compute metrics
metrics_61 = pd.DataFrame({
    "Metric": ["RMSE (GW)", "MAPE (%)"],
    "Value": [
        rmse(comparison_df["Actual"], comparison_df["SARIMA_Forecast"]),
        mape(comparison_df["Actual"], comparison_df["SARIMA_Forecast"])
    ]
})

metrics_61
```

###  Regional Performance Analysis

While the previous sections focused on a single region to illustrate model behavior in detail, electricity demand dynamics differ meaningfully across regions due to climate, population density, industrial activity, and infrastructure characteristics.  
This subsection evaluates forecasting performance across all regions to assess the consistency and robustness of the modeling approach.

Forecast accuracy is computed separately for each region using the same test period and evaluation metrics. This regional comparison helps identify where classical time-series models perform well and where more flexible machine learning approaches may offer additional gains.


```python
from __future__ import annotations

from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
import requests
from shapely.geometry import Point
from statsmodels.tsa.statespace.sarimax import SARIMAX


def download_file(url: str, dest: Path, chunk_size: int = 1024 * 1024) -> Path:
    """Download `url` to `dest` if it does not already exist."""
    dest.parent.mkdir(parents=True, exist_ok=True)
    if dest.exists():
        return dest

    resp = requests.get(url, stream=True, timeout=60)
    resp.raise_for_status()

    with dest.open("wb") as f:
        for chunk in resp.iter_content(chunk_size=chunk_size):
            if chunk:
                f.write(chunk)

    return dest


def first_existing_column(df: pd.DataFrame, candidates: list[str]) -> str:
    """Return the first column in `candidates` that exists in `df`."""
    for col in candidates:
        if col in df.columns:
            return col
    raise ValueError(f"Missing required columns. Tried: {candidates}. Found: {list(df.columns)}")


def rmse(y_true: pd.Series, y_pred: pd.Series) -> float:
    """Root Mean Squared Error."""
    y_true, y_pred = y_true.align(y_pred, join="inner")
    err = y_true - y_pred
    return float((err.pow(2).mean()) ** 0.5)


def mape(y_true: pd.Series, y_pred: pd.Series) -> float:
    """Mean Absolute Percentage Error (percent). Zeros are treated as missing."""
    y_true, y_pred = y_true.align(y_pred, join="inner")
    denom = y_true.abs().replace(0, pd.NA)
    return float(((y_true - y_pred).abs() / denom).mean() * 100)


def fit_sarima_and_score(
    df_features: pd.DataFrame,
    train_data: pd.DataFrame,
    test_data: pd.DataFrame,
    target_col: str = "load_avg_gw",
) -> pd.DataFrame:
    """
    Fit a SARIMA model per region and compute RMSE/MAPE on the test split.

    Expected inputs:
      - df_features: MultiIndex with a "region" level (used to enumerate regions).
      - train_data/test_data: MultiIndex with a "region" level and a datetime index.
      - `target_col`: numeric target column present in train/test.
    """
    forecasts: dict[str, pd.Series] = {}

    regions = df_features.index.get_level_values("region").unique()
    for region in regions:
        y_train = train_data.xs(region, level="region")[target_col].asfreq("MS")
        y_test = test_data.xs(region, level="region")[target_col].asfreq("MS")

        model = SARIMAX(
            y_train,
            order=(1, 1, 1),
            seasonal_order=(1, 1, 1, 12),
            enforce_stationarity=False,
            enforce_invertibility=False,
        )
        results = model.fit(disp=False)

        y_pred = results.forecast(steps=len(y_test))
        y_pred.index = y_test.index
        forecasts[region] = y_pred

    rows: list[dict[str, object]] = []
    for region, y_pred in forecasts.items():
        y_true = test_data.xs(region, level="region")[target_col]
        rows.append(
            {"Region": region, "RMSE (GW)": rmse(y_true, y_pred), "MAPE (%)": mape(y_true, y_pred)}
        )

    return (
        pd.DataFrame(rows)
        .set_index("Region")
        .sort_values("RMSE (GW)")
    )


def build_saudi_basemap() -> gpd.GeoDataFrame:
    """
    Load Saudi Arabia boundary from Natural Earth (Admin 0 – Countries, 110m).

    Note: Natural Earth is a public cartographic dataset and may not reflect official
    boundary positions for all use cases. For official boundaries, use an authoritative source.
    """
    ne_url = (
        "https://naturalearth.s3.amazonaws.com/110m_cultural/"
        "ne_110m_admin_0_countries.zip"
    )
    zip_path = download_file(ne_url, Path("data/ne_110m_admin_0_countries.zip"))
    world = gpd.read_file(zip_path)

    iso_col = first_existing_column(world, ["ISO_A3", "ADM0_A3", "SOV_A3", "GU_A3"])
    sa = world.loc[world[iso_col].astype(str).str.upper().eq("SAU")].copy()

    if sa.empty:
        raise ValueError("Saudi Arabia polygon not found using ISO code 'SAU'.")

    return sa.to_crs(epsg=3857)


def build_region_points() -> gpd.GeoDataFrame:
    """Create illustrative regional reference points and project them for plotting."""
    points = gpd.GeoDataFrame(
        {
            "Region": ["Central", "Eastern", "Western", "Southern"],
            "City": ["Riyadh", "Dammam", "Jeddah", "Abha"],
        },
        geometry=[
            Point(46.6753, 24.7136),  # Riyadh
            Point(50.1033, 26.4207),  # Dammam
            Point(39.1925, 21.4858),  # Jeddah
            Point(42.5117, 18.2164),  # Abha
        ],
        crs="EPSG:4326",
    )
    return points.to_crs(epsg=3857)


def plot_performance_map(
    sa: gpd.GeoDataFrame,
    points_with_metrics: gpd.GeoDataFrame,
    metric: str = "RMSE (GW)",
    title: str | None = None,
) -> None:
    """
    Plot a Saudi Arabia reference map with points colored by a performance metric.

    `metric` should be one of: "RMSE (GW)", "MAPE (%)".
    """
    if metric not in points_with_metrics.columns:
        raise ValueError(f"Metric '{metric}' not found in points GeoDataFrame columns.")

    if title is None:
        title = f"SARIMA Forecast Performance by Region ({metric})"

    fig, ax = plt.subplots(figsize=(9, 7.5))
    ax.set_facecolor("#f5f7fb")

    sa.plot(
        ax=ax,
        edgecolor="#1f2937",
        facecolor="#e5f0ff",
        linewidth=1.2,
        zorder=1,
    )

    points_with_metrics.plot(
        ax=ax,
        column=metric,
        cmap="RdYlGn_r",  # lower is better -> greener
        markersize=140,
        legend=True,
        legend_kwds={"label": metric, "shrink": 0.75, "pad": 0.02},
        edgecolor="white",
        linewidth=1.2,
        zorder=3,
    )

    minx, miny, maxx, maxy = sa.total_bounds
    dx = (maxx - minx) * 0.02
    dy = (maxy - miny) * 0.02

    for geom, region, city, val in zip(
        points_with_metrics.geometry,
        points_with_metrics["Region"],
        points_with_metrics["City"],
        points_with_metrics[metric],
    ):
        value_str = f"{val:.2f}" if metric == "RMSE (GW)" else f"{val:.1f}%"
        ax.text(
            geom.x + dx,
            geom.y + dy,
            f"{region} ({city})\n{metric}: {value_str}",
            fontsize=9,
            bbox=dict(boxstyle="round,pad=0.25", facecolor="white", edgecolor="none", alpha=0.9),
            zorder=4,
        )

    ax.set_title(title, fontsize=13, pad=18)

    ax.set_xlim(minx - (maxx - minx) * 0.06, maxx + (maxx - minx) * 0.06)
    ax.set_ylim(miny - (maxy - miny) * 0.06, maxy + (maxy - miny) * 0.06)

    ax.set_axis_off()

    ax.text(
        0.5,
        0.02,
        "Boundary source: Natural Earth (Admin 0 – Countries, 110m).",
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        fontsize=8,
        color="#374151",
    )

    plt.tight_layout()
    plt.show()


regional_metrics_df = fit_sarima_and_score(df_features, train_data, test_data, target_col="load_avg_gw")
display(regional_metrics_df)

saudi = build_saudi_basemap()
region_points = build_region_points()

region_points = region_points.join(regional_metrics_df, on="Region")
missing = region_points[region_points[["RMSE (GW)", "MAPE (%)"]].isna().any(axis=1)]["Region"].tolist()
if missing:
    raise ValueError(f"Missing metrics for region(s): {missing}. Check naming consistency.")

plot_performance_map(saudi, region_points, metric="RMSE (GW)")
```

```python
# Central region series for visualization
central_series = (
    df_features
    .xs("Central", level="region")["load_avg_gw"]
    .reset_index()
)

central_series["year"] = central_series["date"].dt.year
central_series["month"] = central_series["date"].dt.month

# Pivot to Year x Month matrix (mean load)
heatmap_data = central_series.pivot_table(
    index="year",
    columns="month",
    values="load_avg_gw",
    aggfunc="mean"
).sort_index()

plt.figure(figsize=(12, 6))

# Light red–blue diverging color scheme
plt.imshow(
    heatmap_data,
    aspect="auto",
    cmap="coolwarm"  # alternative: "RdBu_r"
)

plt.colorbar(label="Average Load (GW)")
plt.title("Seasonality Heatmap: Central Region Electricity Load (Year × Month)")
plt.xlabel("Month")
plt.ylabel("Year")

plt.xticks(
    ticks=range(12),
    labels=["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
)
plt.yticks(
    ticks=range(len(heatmap_data.index)),
    labels=heatmap_data.index
)

plt.tight_layout()
plt.show()
```

**Interpretation**

The heatmap highlights a strong and recurring seasonal pattern in electricity demand across the Central region. Load levels rise steadily from spring, peak during the summer months of June through August, and decline toward the end of the year, reflecting temperature-driven cooling demand.

A gradual increase in peak summer intensity over time is also visible, indicating a long-term upward trend in electricity consumption. This combination of strong seasonality and structural growth underscores the importance of explicitly modeling both effects in regional electricity load forecasting.


### Feature Importance and SHAP Analysis

While traditional time-series models emphasize trend and seasonality, machine learning models offer additional interpretability through feature attribution. This subsection examines which engineered features most strongly influence electricity load forecasts and how their effects vary over time.

Two complementary approaches are used:

1. **Global Feature Importance**, which ranks predictors based on their overall contribution to model performance.
2. **SHAP (SHapley Additive exPlanations)**, which provides a unified, game-theoretic framework to explain individual predictions and capture nonlinear feature interactions.

Together, these methods help translate model outputs into insights that are meaningful for energy planning and policy analysis.


```python
import xgboost as xgb

# Feature set
feature_cols = [
    "month",
    "is_summer",
    "load_avg_lag_1",
    "load_avg_lag_3",
    "load_avg_lag_6"
]

# Prepare training data
X_train = train_data[feature_cols]
y_train = train_data["load_avg_gw"]

# Train XGBoost model
xgb_model = xgb.XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

xgb_model.fit(X_train, y_train)

# --- Custom pastel color ---
pastel_blue = "#8FD3FF"  # light blue, consistent with your other visuals

# Plot feature importance
plt.figure(figsize=(8, 5))
ax = xgb.plot_importance(
    xgb_model,
    importance_type="gain",
    show_values=False,
    color=pastel_blue
)

# Improve aesthetics
ax.set_title("XGBoost Feature Importance (Gain)", fontsize=13, pad=10)
ax.set_xlabel("Importance Score", fontsize=11)
ax.set_ylabel("Features", fontsize=11)
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

**Interpretation**

This feature importance chart shows that seasonal effects play a dominant role in forecasting electricity demand, with the summer indicator emerging as the most influential predictor. This reflects the strong impact of temperature-driven cooling demand on regional electricity load. Lagged average load from the previous month also contributes substantially, indicating that recent demand levels provide important information about near-term consumption. Longer lag features and the calendar month have smaller but still meaningful contributions, suggesting that short-term persistence and seasonal structure together shape electricity load dynamics. Overall, the results highlight that electricity demand in Saudi Arabia is driven primarily by seasonal climate effects and recent consumption patterns, reinforcing the importance of incorporating both temporal memory and seasonality into forecasting models.

### Interpreting Regional Differences

Forecasting performance and demand dynamics differ across regions because regional electricity load is shaped by heterogeneous structural drivers. These include climate exposure, population concentration, economic activity, and the intensity of seasonal cooling demand. In this subsection, we connect the earlier error results to observable regional load behavior by comparing (i) average demand levels, (ii) the strength of seasonality, and (iii) demand volatility across regions.

To make these differences interpretable and comparable, we summarize three region-level indicators computed from the full monthly time series:

1. **Mean load (GW)**: the typical level of electricity demand.
2. **Seasonality strength**: the average amplitude of the annual cycle (peak month minus trough month).
3. **Volatility (standard deviation)**: how much demand fluctuates over time.

These descriptive indicators provide an empirical basis for explaining why some regions are easier to forecast than others.


```python
df_summary = (
    df_features
    .reset_index()
    .groupby("region")
    .agg(
        mean_load_gw=("load_avg_gw", "mean"),
        std_load_gw=("load_avg_gw", "std"),
        min_load_gw=("load_avg_gw", "min"),
        max_load_gw=("load_avg_gw", "max"),
    )
)

tmp = df_features.reset_index()
tmp["year"] = tmp["date"].dt.year
tmp["month"] = tmp["date"].dt.month

annual_amp = (
    tmp.groupby(["region", "year"])["load_avg_gw"]
       .agg(lambda s: s.max() - s.min())
       .groupby("region")
       .mean()
       .rename("seasonality_amplitude_gw")
)

df_summary = df_summary.join(annual_amp)

df_region_diagnostics = df_summary.join(regional_metrics_df)

display(df_region_diagnostics.sort_values("RMSE (GW)"))


plot_df = df_region_diagnostics.dropna()

regions = plot_df.index.astype(str)
mean_load = plot_df["mean_load_gw"].to_numpy()
season_amp = plot_df["seasonality_amplitude_gw"].to_numpy()
rmse_vals = plot_df["RMSE (GW)"].to_numpy()
mape_vals = plot_df["MAPE (%)"].to_numpy()


def scale_sizes(x, min_size=80, max_size=450):
    x = np.asarray(x, dtype=float)
    if np.allclose(x.max(), x.min()):
        return np.full_like(x, (min_size + max_size) / 2.0)
    x = (x - x.min()) / (x.max() - x.min())
    return min_size + x * (max_size - min_size)


sizes_by_season = scale_sizes(season_amp)

fig, ax = plt.subplots(figsize=(8.5, 5.5))
ax.scatter(mean_load, rmse_vals, s=sizes_by_season, alpha=0.9, edgecolor="gray", linewidth=0.6)

for x, y, r in zip(mean_load, rmse_vals, regions):
    ax.text(x, y, f" {r}", va="center", fontsize=9)

ax.set_title("Regional Diagnostics: Mean Load vs RMSE\n(Bubble size = seasonality amplitude)")
ax.set_xlabel("Mean Load (GW)")
ax.set_ylabel("RMSE (GW)")
ax.grid(True, linewidth=0.5, alpha=0.3)

plt.tight_layout()
plt.show()


sizes_by_mean = scale_sizes(mean_load)

fig, ax = plt.subplots(figsize=(8.5, 5.5))
ax.scatter(season_amp, mape_vals, s=sizes_by_mean, alpha=0.9, edgecolor="gray", linewidth=0.6)

for x, y, r in zip(season_amp, mape_vals, regions):
    ax.text(x, y, f" {r}", va="center", fontsize=9)

ax.set_title("Regional Diagnostics: Seasonality Amplitude vs MAPE\n(Bubble size = mean load)")
ax.set_xlabel("Seasonality Amplitude (GW)")
ax.set_ylabel("MAPE (%)")
ax.grid(True, linewidth=0.5, alpha=0.3)

plt.tight_layout()
plt.show()


heat_cols = [
    "mean_load_gw",
    "std_load_gw",
    "min_load_gw",
    "max_load_gw",
    "seasonality_amplitude_gw",
    "RMSE (GW)",
    "MAPE (%)",
]

heat_df = df_region_diagnostics[heat_cols].dropna()
z = (heat_df - heat_df.mean()) / heat_df.std(ddof=0)

fig, ax = plt.subplots(figsize=(10.5, 3.8))
im = ax.imshow(z.values, aspect="auto")

ax.set_title("Regional Diagnostics (Standardized Across Metrics)")
ax.set_yticks(range(len(heat_df.index)))
ax.set_yticklabels(heat_df.index)
ax.set_xticks(range(len(heat_cols)))
ax.set_xticklabels(heat_cols, rotation=30, ha="right")

cbar = plt.colorbar(im, ax=ax, shrink=0.9, pad=0.02)
cbar.set_label("z-score")

for i in range(z.shape[0]):
    for j in range(z.shape[1]):
        ax.text(j, i, f"{heat_df.iloc[i, j]:.2f}", ha="center", va="center", fontsize=8)

plt.tight_layout()
plt.show()
```

### Regional Load Characteristics and Forecast Diagnostics

These visuals summarize key regional differences in electricity demand and SARIMA forecast performance. The scatter plots relate average load and seasonality strength to forecast error, highlighting how regions with higher demand levels and stronger seasonal variation tend to exhibit larger absolute and relative errors. ## 8. Policy Implications and Discussion

This section translates the empirical forecasting results into implications for public policy, infrastructure planning, and regional demand management. Electricity load forecasting is not merely a technical exercise in prediction accuracy; it constitutes a core informational input into decisions on generation adequacy, transmission and distribution investment, tariff reform, and the design of demand-side interventions. In Saudi Arabia and the wider Middle East and North Africa region, these decisions are particularly sensitive to peak demand dynamics driven by climate conditions and cooling needs.

The empirical results from both the SARIMA and XGBoost models reinforce a central structural characteristic of the Saudi electricity system: demand is highly seasonal, regionally heterogeneous, and strongly influenced by summer cooling requirements. These features imply that policy relevance lies less in average annual consumption and more in the timing, magnitude, and regional concentration of peak loads. The discussion is organized around three interrelated themes that emerge directly from the modeling results: capacity and infrastructure planning, management of seasonal peak demand, and regional differentiation in policy design.

###  Implications for Capacity and Infrastructure Planning

Regional electricity load forecasts provide a critical bridge between observed demand behavior and infrastructure adequacy planning. In systems characterized by strong seasonality, such as Saudi Arabia’s, the binding constraint for system reliability is often peak adequacy rather than total annual energy supply. The forecasting results show that summer peaks, particularly in the Central and Eastern regions, drive a disproportionate share of system stress and forecasting error. This implies that marginal investments in generation and network capacity are largely justified by a relatively small number of high-load periods.

From a policy perspective, this raises important cost-efficiency considerations. Capacity built to meet extreme summer peaks may remain underutilized for much of the year, increasing long-run system costs. Forward-looking regional studies of cooling demand in Saudi Arabia show that rising temperatures are likely to intensify this challenge over time, particularly in urbanized regions where cooling demand growth is most pronounced (Mikayilov et al., 2025). These findings underscore the importance of integrating regional demand forecasts into capacity planning frameworks rather than relying solely on national aggregates.

Infrastructure planning must also be robust to structural change. The historical dataset used in this study ends in 2019, meaning that the forecasts represent a baseline conditioned on pre-2020 demand structures. Subsequent developments, including electricity price reforms, efficiency standards, renewable energy deployment, and evolving consumption patterns, can alter both the level and shape of electricity demand. In this context, scenario-based forecasting offers a practical extension of the present analysis. Statistical and machine learning forecasts can serve as reference trajectories that are adjusted under alternative policy and technology assumptions, supporting more resilient planning under uncertainty.

### 8.2 Managing Seasonal Peak Demand

The dominant role of summer peaks observed in the empirical analysis points to peak demand management as a central policy objective. The strong explanatory power of seasonal indicators and lagged demand in the XGBoost model highlights the extent to which electricity load is driven by predictable climate-related cycles rather than random fluctuations. This aligns with recent regional research showing that cooling demand is a primary driver of both current peak load and future peak growth in Saudi Arabia (Aldubyan, 2024; Mikayilov et al., 2025).

Peak-oriented demand-side measures are therefore particularly valuable when evaluated in terms of their capacity value rather than solely their annual energy savings. Studies of Saudi residential electricity demand demonstrate that energy efficiency policies, especially those targeting cooling technologies and building performance, can materially reduce peak loads if designed with peak coincidence in mind (Aldubyan, 2024). This supports prioritizing interventions such as high-efficiency air conditioning, building envelope improvements, and cooling control strategies that directly address summer demand.

However, the policy design challenge extends beyond technology adoption. Behavioral responses and price effects play an important role in shaping realized demand outcomes. Evidence from Saudi Arabia indicates that efficiency improvements can be partially offset by rebound effects, particularly in cooling-intensive end uses, where lower effective costs of comfort services lead to increased usage (Belaïd & Mikayilov, 2024). This suggests that peak management strategies are most credible when implemented as integrated policy packages that combine efficiency standards, pricing signals, and targeted demand-side programs rather than relying on isolated instruments.

###  Regional Demand Management Considerations

Regional heterogeneity emerges as a central finding of the empirical analysis rather than a secondary complication. Forecast errors are systematically higher in regions with greater demand volatility and stronger seasonal amplitude, notably the Central region. In contrast, regions with lower and more stable demand profiles, such as the Southern region, exhibit substantially lower forecasting errors. These differences reflect underlying structural drivers, including climate exposure, population density, economic activity, and building stock characteristics.

From a policy perspective, this implies that uniform national interventions may be suboptimal. Regions with pronounced seasonal peaks are likely to yield higher returns from cooling-focused efficiency programs and peak-oriented demand management, while regions with different demand structures may require alternative measures, such as network reinforcement or targeted industrial demand management. Recent Saudi-focused research emphasizes the value of region-specific analysis in understanding electricity demand behavior and policy responsiveness, cautioning against reliance on national averages that mask local dynamics (Mikayilov et al., 2025).

Regional forecasting also has operational value beyond long-term planning. Region-level demand projections can inform transmission planning, maintenance scheduling, congestion risk assessment, and the prioritization of flexibility resources. Institutionalizing regional load forecasting within planning workflows can improve system resilience and reduce the risk of overinvestment driven by uncertainty around peak demand.


## 9. Conclusion and Limitations

This study presented an applied data science framework for forecasting regional electricity load in Saudi Arabia using monthly data from 2009 to 2019. By combining classical time-series modeling with machine learning techniques, the analysis demonstrated the value of regionally disaggregated forecasting for understanding demand dynamics, seasonal peak behavior, and infrastructure planning challenges in a hot-climate electricity system.

### Summary of Key Findings

Several key findings emerge from the empirical analysis.

First, electricity demand in Saudi Arabia exhibits strong and persistent seasonality across all regions, with summer peaks dominating load profiles. This seasonal structure is highly predictable and closely aligned with cooling demand, reinforcing the importance of explicitly modeling seasonal effects in forecasting applications.

Second, meaningful regional heterogeneity exists in both demand levels and volatility. The Central and Eastern regions consistently exhibit higher average load, stronger seasonal amplitude, and greater forecast uncertainty than the Western and Southern regions. These differences are not merely statistical artifacts but reflect underlying structural drivers such as population concentration, economic activity, and climate exposure.

Third, machine learning models, specifically XGBoost, outperform the SARIMA baseline in capturing short-term dynamics and nonlinear relationships, particularly during high-load periods. Feature importance results show that seasonal indicators and recent lagged demand are the most influential predictors, highlighting the dominant role of climate-driven cycles and temporal persistence in shaping electricity demand.

Finally, the results demonstrate that forecasting accuracy is systematically lower in regions with higher volatility and stronger seasonal peaks. This finding has direct implications for policy and planning, as it suggests that the regions where accurate forecasting matters most for system reliability are also those where prediction is most challenging.


### Study Limitations and Data Considerations

Several limitations should be considered when interpreting the findings of this study. First, the analysis relies on historical electricity load data that end in 2019. As a result, the forecasts reflect demand behavior under the institutional, technological, and regulatory conditions prevailing during that period. Subsequent structural changes in the Saudi electricity sector are not explicitly captured, including:

- Electricity price reforms that altered consumer incentives and consumption behavior  
- The introduction and tightening of energy efficiency standards for buildings and appliances  
- Increased penetration of renewable energy and associated changes in generation dispatch  
- Behavioral adjustments by households and firms in response to policy reforms and awareness initiatives  

These developments may affect both the level and temporal structure of electricity demand, particularly peak load dynamics. Consequently, the forecasts should be interpreted as a historical baseline rather than as projections of post-reform demand trajectories.

Second, explicit meteorological variables such as temperature, humidity, or cooling degree days are not included as model inputs. While strong seasonal effects are captured indirectly through calendar-based indicators and lagged demand features, the models learn climate influences implicitly from historical patterns rather than responding directly to contemporaneous weather conditions. As a result, the forecasts are not designed to capture short-term weather shocks or long-term climate-driven shifts beyond those embedded in past demand behavior.

Third, the use of monthly data limits the ability to analyze intraday or intra-week demand dynamics, which are particularly relevant for operational scheduling, peak load management, and short-term reliability assessment. Higher-frequency data would be required to extend the analysis to these use cases.


###  Directions for Future Work

The analysis points to several directions for future research.

A natural extension is the integration of high-frequency temperature and weather data to explicitly model climate sensitivity and extreme heat events. This would allow more precise estimation of cooling-driven peak demand and improve stress-testing of system adequacy under climate change scenarios.

Scenario-based forecasting represents another important avenue. Combining statistical baselines with policy and technology scenarios, such as alternative efficiency trajectories or tariff reforms, would strengthen the policy relevance of the forecasts and support planning under uncertainty.

Future work could also explore hybrid modeling approaches that combine time-series structure with machine learning flexibility, such as state-space models with nonlinear components or recurrent neural networks adapted for regional electricity demand.

Finally, expanding the analysis to include higher-frequency load data and additional regional or sectoral disaggregation would enhance its applicability for operational planning, demand response design, and real-time system management.

Taken together, these extensions would further strengthen the role of data-driven forecasting as a decision-support tool for electricity system planning and energy policy in Saudi Arabia and other hot-climate regions.
